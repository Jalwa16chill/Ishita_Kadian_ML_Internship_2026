```python
# Find area of rectangle.
length = float(input("Enter length of the rectangle: "))
breadth = float(input("Enter breadth of the rectangle: "))
area_rectangle = length * breadth
print(f"Area of Rectangle = {area_rectangle}")
```

```python
# Find simple interest.
principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest: "))
time = float(input("Enter time period (years): "))
simple_interest = (principal * rate * time) / 100  #[cite: 3]
print(f"Simple Interest = {simple_interest}")
```

```python
# Convert temperature from Celsius to Fahrenheit.
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(f"Temperature in Fahrenheit = {fahrenheit}°F")
```

```python
# Calculate average of 3 numbers.
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))
average = (num1 + num2 + num3) / 3
print(f"Average of three numbers = {average}")
```

```python
# Find square and cube of a number.
val = float(input("Enter a number to find its square and cube: "))
square_val = val ** 2  #[cite: 2]
cube_val = val ** 3  #[cite: 2]
print(f"Square of {val} = {square_val}")
print(f"Cube of {val} = {cube_val}")
```

```python
# Swap two numbers without third variable.
x = float(input("Enter first number (x): "))
y = float(input("Enter second number (y): "))
x, y = y, x
print("After swapping processing:")
print(f"x = {x}")
print(f"y = {y}")
```

```python
# Create a Student Report Program that take student details using input(), Store marks in variables, Calculate total and percentage , Use comments, Use proper indentation
st_name = input("Enter Student Name: ")  #[cite: 1]
st_roll = input("Enter Roll Number: ")
st_college = input("Enter College Name: ")

m_python = float(input("Enter marks for Python Foundations: "))
m_ml = float(input("Enter marks for Machine Learning: "))
m_genai = float(input("Enter marks for Generative AI: "))
m_math = float(input("Enter marks for Applied Mathematics: "))
m_stats = float(input("Enter marks for Statistics: "))

total_score = m_python + m_ml + m_genai + m_math + m_stats
final_percentage = (total_score / 500) * 100

print("\n" + "="*40)
print(f"         ACADEMIC TRANSCRIPT REPORT      ")
print("="*40)
print(f"Student Name : {st_name}")
print(f"Roll Number  : {st_roll}")
print(f"Institution  : {st_college}")
print("-"*40)
print(f"Total Marks  : {total_score} / 500")
print(f"Percentage   : {final_percentage:.2f}%")
print("="*40)
```
