```python
# Find sum of first 10 natural numbers.
sum_natural = 0
for i in range(1, 11):
    sum_natural += i
print(f"Sum of first 10 natural numbers = {sum_natural}")
```

```python
# Find factorial of a number.
input_num = int(input("Enter number for factorial: "))  #[cite: 1]
fact_val = 1
for i in range(1, input_num + 1):  #[cite: 2]
    fact_val *= i
print(f"Factorial of {input_num} = {fact_val}")
```

```python
# Print Fibonacci Series.
fib_terms = int(input("Enter terms boundary length for Fibonacci pattern: "))
t1, t2 = 0, 1
print("Fibonacci Series Structure:")
for i in range(fib_terms):  #[cite: 2]
    print(t1, end=" ")
    next_term = t1 + t2
    t1 = t2
    t2 = next_term
print()
```

```python
# Find largest among 3 numbers.
val1 = float(input("Enter number 1: "))
val2 = float(input("Enter number 2: "))
val3 = float(input("Enter number 3: "))

if (val1 >= val2) and (val1 >= val3):  #[cite: 2]
    largest = val1
elif (val2 >= val1) and (val2 >= val3):  #[cite: 2]
    largest = val2
else:
    largest = val3
print(f"The largest entry element = {largest}")
```

```python
# Create Student Result System.
print("\n--- Student Result System Administration Management ---")
st_id = input("Enter Student ID Details: ")
st_name = input("Enter Student Full Name: ")

num_subjects = 3
total_possible_marks = num_subjects * 100
aggregated_score = 0

for idx in range(1, num_subjects + 1):
    subject_mark = float(input(f"Enter entry validation marks for Subject {idx}: "))
    aggregated_score += subject_mark

score_percentage = (aggregated_score / total_possible_marks) * 100

if score_percentage >= 90:
    assigned_letter_grade = "A"
elif score_percentage >= 75:
    assigned_letter_grade = "B"
elif score_percentage >= 50:
    assigned_letter_grade = "C"
else:
    assigned_letter_grade = "Fail"

print("\n" + "="*45)
print("             METRIC EVALUATION ANALYSIS        ")
print("="*45)
print(f"Identification Target : {st_id}")
print(f"Student Trainee Label : {st_name}")
print(f"Aggregated Profile    : {aggregated_score:.1f} / {total_possible_marks}")
print(f"Calculated Metric     : {score_percentage:.2f}%")
print(f"Final Assigned Grade  : {assigned_letter_grade}")
print("="*45)
```
